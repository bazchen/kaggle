## 4.
What are the most common wine-producing regions? Create a Series counting the number of times each value occurs in the `region_1` field. This field is often missing data, so replace missing values with `Unknown`. Sort in descending order.  Your output should look something like this:

```
Unknown                    21247
Napa Valley                 4480
                           ...  
Bardolino Superiore            1
Primitivo del Tarantino        1
Name: region_1, Length: 1230, dtype: int64
```

````python
# 1 clean data. add unknown to missing values
cleaned_reviews = reviews.region_1.fillna("Unknown")

# 2 group by wine-producing region and count
grouped_regions_counts = cleaned_reviews.groupby(cleaned_reviews).value_counts()

# 3 sort in descending ordder
reviews_per_region = grouped_regions_counts.sort_values(ascending=False)

# Check your answer
q4.check()

reviews_per_region
````
